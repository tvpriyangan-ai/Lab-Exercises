```python
def count_even_number(numbers):
    count = 0
    for value in numbers:
        if value % 2 == 0:
            count += 1
    return count

    count_even_number=([2,3,5,6,7])

```

```python
def find_maximum(numbers):
  maximum=numbers[0]
  for i in numbers:
    if i>=maximum:
      i=maximum
      return maximum

  find_maximum=([3,4,5,6,7])
```

```python
def total_length(lst):
  total=0
  for i in lst:
    total+=len(lst)
  return total

  total_length=(["LAPTOP", "PHONE", "PC"])
```

```python
def shortest_one(numbers):
  index=0
  for i in range(1, len(numbers)):
    if len(numbers[i])<len(numbers[index]):
      index=i
    return numbers[index]

    shortest_one=(["Books", "Electronics", "Handicrafts"])

```
