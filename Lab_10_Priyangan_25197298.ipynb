## Sorting Algorithm Analysis

Here are the sorting algorithms we'll be using in the lab:

```python
# Sorting Algorithms.

def bubble_sort(arr):
    n = len(arr)

    ## Traverse through all array elements
    swaps = 1
    while swaps>0:
        #print('Sweeping through list')
        swaps = 0
        ## Last i elements are already in place
        for j in range(0, n-1):
            ## traverse the array from 0 to n-i-1
            ## Swap if the element found is greater
            ## than the next element
            if arr[j] > arr[j+1] :
                #print('Swapping ',arr[j],' and ',arr[j+1])
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swaps+=1

        #if swaps==0:
            #print('No more swaps found\n')




def insertion_sort(lst):
    n = len(lst)
    for i in range(n):
        j = i
        while j > 0 and lst[j-1] > lst[j]:
            # swap the current value one space left
            tmp = lst[j]
            lst[j] = lst[j-1]
            lst[j-1] = tmp
            j -= 1

def merge_sort(lst):
    # if the list is this short, no sorting is necessary
    if len(lst) <= 1:
        # return a copy, out-of-place
        return list(lst)

    # half the length, rounded down
    m = len(lst) // 2

    # list slicing
    A = lst[:m]
    B = lst[m:]

    # sort both halves recursively
    A = merge_sort(A)
    B = merge_sort(B)

    return linear_merge(A, B)

def linear_merge(A, B):
    # merge two sorted lists
    output = []
    i, j = 0, 0

    while i + j < len(A) + len(B):
        if i < len(A) and (j == len(B) or A[i] <= B[j]):
            output.append(A[i])
            i += 1
        else:
            output.append(B[j])
            j += 1

    return output
```

Create a list with the values [64, 34, 25, 12, 22, 11, 90], sort the values with each algorithm and print out how long each took. Remember to not hand the algorithms an already sorted version of the array!

```python
# values in list:
arr = [64, 34, 25, 12, 22, 11, 90]

# put your code here.


bubble_arr = arr.copy()
start = time.time()
bubble_sort(bubble_arr)
bubble_time = time.time() - start
print(f"Bubble Sort Time: {bubble_time * 1_000_000: .2f} microseconds")

insertion_arr = arr.copy()
start = time.time()
insertion_sort(insertion_arr)
insertion_time = time.time() - start
print(f"Insertion Sort Time: {insertion_time * 1_000_000: .2f} microseconds")

merge_arr = arr.copy()
start = time.time()
merge_sort(merge_arr)
merge_time = time.time() - start
print(f"Merge Sort Time: {merge_time*1_000_000: .2f} microseconds")
```

The above array was very small and sorting is trivial. Now extend this by sorting a list of shuffled integers from 0 to 999. and print out the results for each algorithm.

```python
# put your code here.

import time
import random

new_list = list(range(1000))
random.shuffle(new_list)

bubble_new_list=new_list.copy()
start=time.time()
bubble_sort(bubble_new_list)
bubble_time=time.time()-start
print(f"The bubble sort time:{bubble_time*1_000: .2f} Milliseconds")

insertion_new_list=new_list.copy()
start=time.time()
insertion_sort(insertion_new_list)
insertion_time=time.time()-start
print(f"The Insertion Time:{insertion_time*1_000: .2f} Milliseconds")

merge_new_list=new_list.copy()
start=time.time()
merge_sort(merge_new_list)
merge_time=time.time()-start
print(f"The merge time:{merge_time*1_000: .2f} Milliseconds")

```

In order to be robust with Dynamic Analysis the results should always be averaged over several runs. Again, sort the numbers 0-999 which have been shuffled, but repeat the analysis 100 times to get an average for comparison. How much does this vary from the above single-run results?
